```python
import pandas as pd
import requests
import os

# Standard Conceptual Captions TSVs often lack headers, so we name them manually
df = pd.read_csv('CC_TSV/Train_GCC-training.tsv', sep='\t', names=['caption', 'url'])

output_dir = "Dataset_CC"
os.makedirs(output_dir, exist_ok=True)

for index, row in df.iterrows():
    try:
        # Timeout is crucial as many links will hang
        img_data = requests.get(row['url'], timeout=10).content
        
        # Save image using the index as the filename
        with open(os.path.join(output_dir, f"{index}.jpg"), 'wb') as handler:
            handler.write(img_data)
    except Exception as e:
        print(f"Failed to download {row['url']}: {e}")
```
